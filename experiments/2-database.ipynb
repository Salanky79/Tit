
Database cần lưu:
- Danh sách các cuộc hội thoại
- Các tin nhắn trong từng câu hội thoại
- Các thông tin cần được ghi nhớ (memory) của agent


Bảng 1 — conversations
Với mỗi cuộc chat, muốn biết đó là cuộc chat nào, tên gì và lần cuối cập nhật lúc nào.

Vậy ta thiết kế các cột:
Cột	            Lý do cần có
id	            Mã số của bản ghi
thread_id	    Mã định danh cuộc chat mà ứng dụng sử dụng
title	        Tiêu đề hiển thị trên giao diện
created_at	    Thời điểm tạo cuộc chat
updated_at	    Thời điểm cập nhật gần nhất


Bảng 2 — chat_messages
Một cuộc chat có thể có rất nhiều tin nhắn. Vì vậy, mỗi tin nhắn cần được lưu thành một bản ghi riêng.
Cột	            Lý do cần có
id	            Mã bản ghi tin nhắn
thread_id	    Tin nhắn thuộc cuộc chat nào
role	        Ai gửi: user hay assistant
content	        Nội dung tin nhắn
created_at	    Thời điểm tạo tin nhắn


Bảng 3 — long_term_memory
Tương tự, mỗi thông tin ghi nhớ là một bản ghi riêng.
Cột	            Lý do cần có
id	            Mã bản ghi
thread_id	    Bộ nhớ thuộc cuộc chat nào
memory	        Nội dung cần ghi nhớ
created_at	    Thời điểm tạo bộ nhớ
